```python
    
import numpy as np

import spacy
import nltk
import re
from nltk.corpus import stopwords
from collections import Counter 
```

```python
def clean_text(text):
    if not isinstance(text, str):
        return ""
    
    # Lowercase
    text = text.lower()
    
    # Remove special characters and digits
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    
    # Remove extra whitespaces
    text = " ".join(text.split())
    
    return text
```

```python
# NLP Cleaning: 
# 1. Normalization (lowercase)
# 2. Punctuation removal
# 3. Tokenization
# 4. Stopwords removal

def nlp_cleaning(text):
    if not isinstance(text, str):
        return ""
    
    # 1. Normalization
    text = text.lower()
    
    # 2. Punctuation removal (anything not alphanumeric or space)
    text = re.sub(r'[^a-z0-9\s]', '', text)
    
    # 3. Tokenization & 4. Stopwords removal with SpaCy
    doc = nlp(text)
    # Keep tokens that are not stopwords and not empty
    tokens = [token.text for token in doc if not token.is_stop and token.text.strip()]
    
    return " ".join(tokens)

# # Test example
# sample_text = "Hello World! THIS is a simple test with punctuation, numbers 123, and stopwords."
# print(f"Original: {sample_text}")
# print(f"Cleaned:  {nlp_cleaning(sample_text)}")
```
